```python
import matplotlib.pyplot as plt
import networkx as nx


class FSA:
    def __init__(
        self, states, alphabet, transition_function, start_state, accept_states
    ):
        self.states = states
        self.alphabet = alphabet
        self.transition_function = transition_function
        self.start_state = start_state
        self.accept_states = accept_states

    def simulate(self, input_string):
        current_state = self.start_state
        path = [current_state]

        for symbol in input_string:
            if symbol not in self.alphabet:
                return False, path + ["invalid symbol"]

            if (current_state, symbol) in self.transition_function:
                current_state = self.transition_function[(current_state, symbol)]
                path.append(current_state)
            else:
                return False, path + ["no transition"]
        return current_state in self.accept_states, path
    
    def visualize(self):
        GRP = nx.DiGraph()

        for state in self.states:
            if state == self.start_state:
                GRP.add_node(state, color="lightgreen")
            elif state in self.accept_states:
                GRP.add_node(state, color="lightblue")
            else:
                GRP.add_node(state, color="lightgray")

        for (from_state, symbol), to_state in self.transition_function.items():
            GRP.add_edge(from_state, to_state, label=symbol)

        pos = nx.spring_layout(GRP) 
        node_colors = [GRP.nodes[n]["color"] for n in GRP.nodes()]

        nx.draw(GRP, pos, with_label=True, node_colors=node_colors, node_size=2000, edge_color="black", width=1, arrow_style="->", arrowsize=20)

        edge_labels = nx.get_edge_attributes(GRP, "label")
        nx.draw_networkx_edge_labels(GRP, pos, edge_labels=edge_labels)

        plt.title("Diagram FSA")
        plt.show()

if __name__ == "__main__":
    states = {"q0", "q1", "q2"}
    alphabet = {"a", "b"}
    transition_function = {
        ("q0", "a"): "q1",
        ("q0", "b"): "q0",
        ("q1", "a"): "q1",
        ("q1", "b"): "q2",
        ("q2", "a"): "q2",
        ("q2", "b"): "q2"
    }

    start_state = "q0"
    accept_states = {"q2"}

    fsa = FSA(states, alphabet, transition_function, start_state, accept_states)
    fsa.visualize()

    input_string = input("Masukkan string: ")
    is_accepted, path = fsa.simulate(input_string)
    
    print(f"\nHasil simulasi: ")
    print(f"Input: {input_string}")
    print(f"Diterima: {is_accepted}")
    print(f"Path: {'-> '.join(path)}")
```

```python
import pandas as pd

df = pd.read_excel("Datasets Superstore.xlsx")
df.head()
```

```python
from sklearn.neighbors import KNeighborsRegressor

latih_data = df.sample(frac=0.8, random_state=1)
uji_data = df.drop(latih_data.index)

model = KNeighborsRegressor(n_neighbors=5)
model.fit(latih_data[['Sales', 'Discount', 'Quantity']], latih_data['Sales'])

prediksi = model.predict(uji_data[['Sales', 'Discount', 'Quantity']])

from sklearn.metrics import mean_squared_error
mse = mean_squared_error(uji_data[['Quantity']], prediksi)
print("Mean Squared Error:", mse)
```

```python
import pandas as pd
from keras.models import Sequential
from keras.layers import Dense

latih_data = df.sample(frac=0.8, random_state=1)
uji_data = df.drop(latih_data.index)

model = Sequential()
model.add(Dense(10, activation='relu'))
model.add(Dense(32, activation='relu'))
model.add(Dense(1, activation='linear'))

model.compile(loss='mean_squared_error', optimizer='adam')
model.fit(latih_data[['Sales', 'Discount', 'Quantity']], latih_data['Sales'], epochs=10, validation_split=0.2)

prediksi = model.predict(uji_data[['Sales', 'Discount', 'Quantity']])

from sklearn.metrics import mean_squared_error
mse = mean_squared_error(uji_data['Discount'], prediksi)
print("Mean Squared Error:", mse)
```
